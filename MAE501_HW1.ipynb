```python
#creates a new object W
#each W object will represent an element in the vector space
class W:
    # W is the set of positive real numbers
    # Vector addition is defined as x + y = xy
    # Scalar multiplication is defined as c*x = x**c

#_init_ runs automatically when a new W object is created
#self is the W object we are creating
    def __init__(self, value):
      #W only has positive real numbers
        if value <= 0:
            raise ValueError("Elements of W must be positive real numbers.")
        self.value = value

#_add_ tells python what to do when it sees x+y
    def __add__(self, other):
      #self.value is number in x and other.value is number in y
      #multiply numbers to get new W object with the answer
        return W(self.value * other.value)

#_rmul_ tells python what to do with a scalar multiplication
#** means to the power of
    def __rmul__(self, scalar):
        return W(self.value ** scalar)

#_repr_ tells python how i want object to be displayed
    def __repr__(self):
      #f lets me put the self.value in the text so itll show W(6) or something
        return f"W({self.value})"
```

```python
# ---------------- TESTING THE CLASS ----------------

# Create two elements of W
x = W(2)
y = W(3)

# Test addition
print(x + y)        # Should give W(6), because 2 * 3 = 6

# Test scalar multiplication
print(3 * x)        # Should give W(8), because 2^3 = 8

# Test with a negative scalar
print(-1 * x)       # Should give W(0.5), because 2^(-1) = 1/2

# Test with a zero scalar
print(0 * x)        # Should give W(1), because 2^0 = 1

# Test another addition
a = W(4)
b = W(5)
print(a + b)        # Should give W(20), because 4 * 5 = 20
```
